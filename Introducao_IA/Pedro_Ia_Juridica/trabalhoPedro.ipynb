## Problema



Muitas pessoas possuem dúvidas jurídicas e não sabem qual área do Direito está relacionada ao seu problema.

O acesso inicial à informação jurídica pode ser difícil para quem não possui formação na área.

Este projeto busca auxiliar na identificação preliminar de casos jurídicos por meio de Inteligência Artificial.

## Objetivo



Desenvolver um agente inteligente capaz de:

- Classificar casos jurídicos.
- Identificar áreas do Direito.
- Buscar jurisprudências relacionadas.
- Gerar recomendações.
- Explicar a decisão tomada.

## Onde a Inteligência Artificial é utilizada

O modelo de linguagem (LLM) executado localmente através do Ollama é utilizado em duas etapas do agente.

1. Classificação do caso

Quando o usuário descreve um problema jurídico, o texto é enviado ao modelo de linguagem.

O modelo analisa o conteúdo e identifica a área jurídica mais compatível, como Direito do Consumidor, Trabalhista, Civil, Família ou Previdenciário.

Exemplo:

Entrada:
"Comprei um produto e ele nunca foi entregue."

Saída:
"Consumidor"

Essa classificação é utilizada nas etapas seguintes do sistema.

2. Explicação da decisão

Após a classificação, o sistema consulta sua base de conhecimento e gera uma recomendação.

Em seguida, essas informações são enviadas novamente ao modelo de linguagem para produzir uma explicação em linguagem natural.

Dessa forma, o usuário não recebe apenas uma resposta, mas também uma justificativa compreensível sobre o motivo daquela recomendação.

O LLM não substitui as regras do sistema. Ele atua como um componente de interpretação e comunicação dentro do fluxo do agente.

## Padrões de IA Agêntica Utilizados

O modelo Gemma 3 4B é utilizado em duas etapas:

1. Classificação jurídica do caso informado pelo usuário.

2. Geração de explicação sobre a decisão tomada pelo sistema.

Dessa forma, o LLM atua como mecanismo de interpretação textual e apoio à tomada de decisão do agente.

### Prompt Chaining

O sistema executa uma sequência de etapas encadeadas.

* 1.  Recebe a descrição do caso.
* 2.  O LLM classifica a área jurídica.
* 3.  O sistema consulta a base de jurisprudências.
* 4.  O sistema gera uma recomendação.
* 5.  O LLM produz uma explicação da decisão.

Cada etapa depende da saída da etapa anterior.

### Tool Use

O agente utiliza ferramentas implementadas em Python.

Exemplos:

* buscar_jurisprudencia()
* gerar_recomendacao()
* calcular_confianca()

Essas funções auxiliam o processo de tomada de decisão.

### Routing

Após a classificação do caso, o sistema escolhe qual conjunto de informações utilizar com base na área jurídica identificada.

### Human-in-the-Loop

O usuário fornece a descrição do caso e pode avaliar criticamente a recomendação produzida pelo sistema.

## Classificação Jurídica

````python
from ollama import chat

def classificar_area(caso):

    prompt = f"""
Você é um classificador jurídico.

Classifique o caso em APENAS UMA destas categorias:

Consumidor
Trabalhista
Civil
Família
Previdenciário

Regras:

- compra, venda, entrega, garantia, cobrança, produto e serviço = Consumidor

- emprego, salário, demissão, patrão e horas extras = Trabalhista

- pensão alimentícia, guarda, casamento, divórcio e filhos = Família

- aposentadoria, INSS, benefício e auxílio-doença = Previdenciário

- contratos, indenização, responsabilidade civil e danos = Civil

Caso:

{caso}

Responda SOMENTE com o nome exato da categoria.
"""

    resposta = chat(
        model='gemma3:4b',
        messages=[
            {
                'role': 'user',
                'content': prompt
            }
        ]
    )

    texto = resposta['message']['content'].strip().lower()

    print("RESPOSTA BRUTA:", repr(texto))

    if "consumidor" in texto:
        return "Consumidor"

    elif "trabalh" in texto:
        return "Trabalhista"

    elif "fam" in texto:
        return "Família"

    elif "previd" in texto:
        return "Previdenciário"

    elif "civil" in texto:
        return "Civil"

    return "Desconhecido"
````

````python
def calcular_confianca(caso, area):

    texto = caso.lower()

    palavras = {

    "Consumidor": [
        "produto",
        "compra",
        "comprado",
        "entrega",
        "loja",
        "garantia",
        "cartão",
        "cobrança",
        "cliente",
        "pedido",
        "serviço",
        "empresa"
    ],

    "Trabalhista": [
        "emprego",
        "demissão",
        "salário",
        "patrão",
        "horas extras",
        "trabalho",
        "empresa",
        "funcionário",
        "empregado",
        "férias",
        "fgts"
    ],

    "Família": [
        "filho",
        "filhos",
        "pensão",
        "guarda",
        "casamento",
        "divórcio",
        "pai",
        "mãe",
        "família"
    ],

    "Civil": [
        "contrato",
        "acordo",
        "indenização",
        "danos",
        "prejuízo",
        "responsabilidade",
        "descumpriu",
        "quebra"
    ],

    "Previdenciário": [
        "aposentadoria",
        "inss",
        "benefício",
        "auxílio",
        "auxílio-doença",
        "segurado",
        "perícia"
    ]
}

    pontos = 0

    for palavra in palavras.get(area, []):

        if palavra in texto:
            pontos += 1

    confianca = 50 + (pontos * 10)

    print("DEBUG PONTOS:", pontos)

    return min(confianca, 95)
````
